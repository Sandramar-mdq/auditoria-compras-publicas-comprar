### 4.4. Formatos e Inconsistencias en CUITs de Proveedores
* CUITs no estandarizados: Al validar el formato estándar (XX-XXXXXXXX-X), se identificaron 2.597 registros con CUITs atípicos:

* Proveedores extranjeros codificados con números cortos o identificadores fiscales internacionales (ej. EMI941031U22, 13-3830415).

* Códigos de identificación interna provisorios asignados por el sistema COMPR.AR (ej. 887, 581).

* CUITs con dígitos sobrantes por error de tipeo administrativo (ej. 30-656088636-6)


```python
# CUITs con formato atípico o internacional
regex_cuit = '^[0-9]{2}-[0-9]{8}-[0-9]{1}$'
cuits_atipicos = df[~df['CUIT'].str.contains(regex_cuit, regex=True)]

print(f"Cantidad de CUITs fuera del formato estándar: {len(cuits_atipicos)}")
print("\nEjemplos de CUITs atípicos o proveedores internacionales:")
print(cuits_atipicos[['CUIT', 'Descripcion_Proveedor']].head(10))
```

### 4.5. Rangos de Valores y Outliers (Valores Atípicos)

* **Outliers en Montos:** El valor mínimo es $0,00 y el máximo asciende a $2.552.449.000,00 (2,55 miles de millones de pesos). El 99% de los montos adjudicados son inferiores a $39.960.000. ### 6. Limpieza y Transformación de Datos

En esta etapa realizamos las modificaciones necesarias para preparar el dataset, garantizando la consistencia de los tipos de datos y dejando el dataframe listo para el análisis cuantitativo y temporal.

### Operaciones realizadas y justificación:
1. **Tratamiento de Valores Faltantes:**
   * `Apartado_Directa`: Rellenado con `"No aplica"`, ya que los valores nulos corresponden a procesos que no son contrataciones directas.
   * `Modalidad`: Rellenado con `"Sin Especificar"` para eliminar el registro nulo puntual.
   ---
2. **Modificación de Tipos de Datos (Conversión Temporal):**
   * Conversión de `Fecha_de_Adjudicacion` y `Fecha_de_perfeccionamiento_OC` a tipo `datetime` mediante `pd.to_datetime()`.
   ---
3. **Normalización de Textos:**
   * Eliminación de espacios sobrantes y del delimitador `;` al final de la columna `Rubros` (`.str.rstrip(';')`).
   * Estandarización de nombres de proveedores (`Descripcion_Proveedor`) a mayúsculas sin espacios adicionales.
   ---
4. **Creación de Nuevas Variables (Ingeniería de Características):**
   * Extracción de `Anio_OC` y `Mes_OC` a partir de la fecha de perfeccionamiento de la Orden de Compra para facilitar las agrupaciones temporales.
   * Extracción de `Anio_Adjudicacion` para análisis comparativos.
   * Creación del indicador booleano `Es_Monto_Valido` para identificar transacciones mayores a cero ($0,00).

```python
# Crear una copia de trabajo para el proceso de limpieza
df_limpio = df.copy()

# 1. Tratamiento de valores faltantes
df_limpio['Apartado_Directa'] = df_limpio['Apartado_Directa'].fillna('No aplica')
df_limpio['Modalidad'] = df_limpio['Modalidad'].fillna('Sin Especificar')

# 2. Modificación de tipos de datos (Conversión a Datetime)
df_limpio['Fecha_de_Adjudicacion'] = pd.to_datetime(df_limpio['Fecha_de_Adjudicacion'], errors='coerce')
df_limpio['Fecha_de_perfeccionamiento_OC'] = pd.to_datetime(df_limpio['Fecha_de_perfeccionamiento_OC'], errors='coerce')

# 3. Normalización de cadenas de texto
df_limpio['Rubros'] = df_limpio['Rubros'].astype(str).str.rstrip(';').str.strip()
df_limpio['Descripcion_Proveedor'] = df_limpio['Descripcion_Proveedor'].astype(str).str.strip().str.upper()
df_limpio['Descripcion_SAF'] = df_limpio['Descripcion_SAF'].astype(str).str.strip()
df_limpio['Descripcion_UOC'] = df_limpio['Descripcion_UOC'].astype(str).str.strip()

# 4. Creación de nuevas variables para análisis
df_limpio['Anio_OC'] = df_limpio['Fecha_de_perfeccionamiento_OC'].dt.year
df_limpio['Mes_OC'] = df_limpio['Fecha_de_perfeccionamiento_OC'].dt.month
df_limpio['Anio_Adjudicacion'] = df_limpio['Fecha_de_Adjudicacion'].dt.year
df_limpio['Es_Monto_Valido'] = df_limpio['Monto'] > 0

# Verificación general del dataframe resultante
print("--- Resumen de la estructura del dataset transformado ---")
df_limpio.info()

# Optimización de memoria (Conversión a tipo categórico)
cols_categoricas = ['Tipo_de_Procedimiento', 'Moneda', 'Modalidad', 'Tipo_de_Operacion']
for col in cols_categoricas:
    df_limpio[col] = df_limpio[col].astype('category')
```
